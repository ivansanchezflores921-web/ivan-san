```python
# Taller: Visualizar Función Convexa

import numpy as np
import matplotlib.pyplot as plt

# 1. Definir la función convexa
def f(x):
    return x**2 + 2*x + 1

# 2. Crear rango de valores
x = np.linspace(-4, 2, 100)
y = f(x)

# 3. Graficar
plt.plot(x, y, label='f(x) = x**2 + 2x + 1')
plt.scatter(-1, 0, color='red', label='Mínimo (-1, 0)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definir las funciones
def f1(x):
    return x**2

def f2(x):
    return x**4 - 5*(x**2)

# 2. Crear rango de valores
x = np.linspace(-3, 3, 100)
y1 = f1(x)
y2 = f2(x)

# 3. Graficar
plt.plot(x, y1, label='f1(x) = x^2 (Convexa)')
plt.plot(x, y2, label='f2(x) = x^4 - 5x^2 (No convexa)')

plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definir la función de costo
def J(w):
    return (w - 3)**2

# 2. Crear rango de valores para el parámetro w
w = np.linspace(0, 6, 100)
costo = J(w)

# 3. Graficar
plt.plot(w, costo, label='J(w) = (w - 3)^2')
plt.scatter(3, 0, color='green', label='w óptimo (3, 0)')

plt.xlabel('w')
plt.ylabel('J(w)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#codigo del dia 10-09-2026
import numpy as np
import matplotlib.pyplot as plt

# Configuración del circuito RC
R = 10000          # 10 kohms
C = 50e-6          # 50 microFaradios
tau = R * C        # tau = 0.5 segundos
Vi = 5.0           # Voltaje de entrada

# Vector de tiempo (de 0 a 5 tau)
t = np.linspace(0, 5 * tau, 1000)

# Ecuación de carga del capacitor: Vc(t) = Vi * (1 - e^(-t / tau))
Vc = Vi * (1 - np.exp(-t / tau))

# Graficar respuesta transitoria
plt.figure(figsize=(6, 3))
plt.plot(t, Vc, label='Vc(t)', color='#D47A3A', lw=2)
plt.axvline(tau, color='#5A6B7C', ls='--', label='tau = 0.5s')
plt.xlabel('Tiempo (s)')
plt.ylabel('Voltaje (V)')
plt.title('Carga de Capacitor en Circuito RC')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Voltaje de entrada
Vi = 5.0

# Caso 1: R = 10 kohms, C = 50 microFaradios
R1 = 10000
C1 = 50e-6
tau1 = R1 * C1

# Caso 2: Aumentamos R
R2 = 20000
C2 = 50e-6
tau2 = R2 * C2

# Caso 3: Aumentamos C
R3 = 10000
C3 = 100e-6
tau3 = R3 * C3

# Vector de tiempo
t = np.linspace(0, 5, 1000)

# Ecuaciones de carga
Vc1 = Vi * (1 - np.exp(-t / tau1))
Vc2 = Vi * (1 - np.exp(-t / tau2))
Vc3 = Vi * (1 - np.exp(-t / tau3))

# Graficar
plt.figure(figsize=(8, 5))

plt.plot(t, Vc1, label='R=10kΩ, C=50µF')
plt.plot(t, Vc2, label='R=20kΩ, C=50µF')
plt.plot(t, Vc3, label='R=10kΩ, C=100µF')

plt.xlabel('Tiempo (s)')
plt.ylabel('Voltaje (V)')
plt.title('Comparación de la carga del capacitor')
plt.legend()
plt.grid(True)

plt.show()
```

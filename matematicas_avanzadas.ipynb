```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. Generar datos de prueba aleatorios
# ==========================================
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
# Relación real: y = 3x + 4 + ruido
y = 4 + 3 * X + np.random.randn(100, 1)

# ==========================================
# 2. Inicialización de parámetros
# ==========================================
w = np.random.randn(1)[0] # Pendiente inicial aleatoria
b = np.random.randn(1)[0] # Sesgo inicial aleatorio

learning_rate = 0.1
iterations = 100
n = len(X)

# Listas para guardar el historial y visualizar después
loss_history = []
w_history = []
b_history = []

w_inicial, b_inicial = w, b # Guardamos para la gráfica de la recta inicial

# ==========================================
# 3. Optimización sin Scikit-Learn
# ==========================================
for i in range(iterations):
    # Predicción del modelo lineal: y^ = wx + b
    y_pred = w * X + b

    # Calcular Función de Pérdida (MSE)
    mse = (1/n) * np.sum((y - y_pred)**2)
    loss_history.append(mse)

    # Calcular Derivadas Parciales
    dw = -(2/n) * np.sum(X * (y - y_pred))
    db = -(2/n) * np.sum(y - y_pred)

    # Actualizar parámetros simultáneamente
    w = w - learning_rate * dw
    b = b - learning_rate * db

    w_history.append(w)
    b_history.append(b)

print(f"Parámetros finales -> Pendiente (w): {w:.4f}, Sesgo (b): {b:.4f}")

# ==========================================
# 4. Visualización del Entrenamiento
# ==========================================
plt.figure(figsize=(14, 5))

# Gráfico 1: Evolución de la recta
plt.subplot(1, 2, 1)
plt.scatter(X, y, color='blue', label='Datos originales', alpha=0.5)

# Recta inicial (roja y punteada)
plt.plot(X, w_inicial * X + b_inicial, color='red', linestyle='--', label='Recta Inicial')

# Algunas rectas intermedias para ver la evolución
for i in range(10, iterations, 20):
    plt.plot(X, w_history[i] * X + b_history[i], color='gray', alpha=0.3)

# Recta final (verde)
plt.plot(X, w * X + b, color='green', linewidth=2, label='Recta Final Ajustada')

plt.title('Evolución de la Recta de Regresión')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()

# Gráfico 2: Caída del Error (Loss)
plt.subplot(1, 2, 2)
plt.plot(range(iterations), loss_history, color='purple', linewidth=2)
plt.title('Caída del Error (MSE) vs Iteraciones')
plt.xlabel('Iteraciones')
plt.ylabel('Pérdida (MSE)')
plt.grid(True)

plt.tight_layout()
plt.show()
print("Valor inicial de b:", b)
```

```python
import matplotlib.pyplot as plt

# Función: f(x) = x²
def funcion(x):
    return x**2


# Algoritmo de descenso por gradiente
def gradiente_descendiente(x, learning_rate, iteraciones):

    historia = []

    for i in range(iteraciones):

        # Gradiente de f(x) = x²
        gradiente = 2 * x

        # Guardar el valor actual
        historia.append(x)

        # Actualizar x
        x = x - learning_rate * gradiente

    return x, historia

# Valor inicial
x_inicial = 10

# Tasas de aprendizaje
tasas_aprendizaje = [0.001, 0.01, 0.1, 0.5, 1.0]

# Número de iteraciones
iteraciones = 20

for tasa in tasas_aprendizaje:

    resultado, historia = gradiente_descendiente(
        x_inicial,
        tasa,
        iteraciones
    )

    print("Tasa de aprendizaje:", tasa)
    print("Valor inicial:", x_inicial)
    print("Valor final:", resultado)
    print("----------------------------------------")

plt.figure(figsize=(10, 6))

for tasa in tasas_aprendizaje:

    resultado, historia = gradiente_descendiente(
        x_inicial,
        tasa,
        iteraciones
    )

    plt.plot(
        range(iteraciones),
        historia,
        marker='o',
        label=f"Tasa = {tasa}"
    )


plt.axhline(0, linestyle='--')

plt.title("Descenso por Gradiente")
plt.xlabel("Iteraciones")
plt.ylabel("Valor de X")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Definimos la función
def f(x):
    return x**2 - 6*x + 13

# Definimos la derivada
def gradiente(x):
    return 2*x - 6

# Parámetros
x = 10              # Valor inicial
alpha = 0.1         # Tasa de aprendizaje
iteraciones = 50    # Número de iteraciones

# Descenso por gradiente
for i in range(iteraciones):
    x = x - alpha * gradiente(x)

    print(f"Iteración {i+1}: x = {x:.6f}, f(x) = {f(x):.6f}")

# Resultado final
print("\nResultado final:")
print(f"x mínimo = {x:.6f}")
print(f"Valor mínimo de f(x) = {f(x):.6f}")
```

```python
import numpy as np
from scipy.optimize import linprog

# Coeficientes de la func. objetivo (-Z)
c = [-3, -2]  # para maximizar Z = 3x1 + 2x2

# Matriz de restricciones de desigualdad (A_ub)
A = [
    [1, 1],  # x1 + x2 <= 4
    [2, 1]   # 2x1 + x2 <= 5
]

# Vector de limites superiores (b_ub)
b = [4, 5]

# Resolver problema lineal (X1>=0, X2>=0)
resultado = linprog(
    c,
    A_ub=A, b_ub=b,
    bounds=(0, None),  # así como en la foto, aplica a las dos variables
    method='highs'
)

print('Solucion Optima:', resultado.x) # te da [1. 3.]
print('Beneficio Maximo Z:', -resultado.fun) # IMPORTANTE: con - porque fun viene negativo
```

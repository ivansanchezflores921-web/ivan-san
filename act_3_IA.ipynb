```python

class AgenteLimpiador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.nivel_bateria = 100
        self.habitacion_actual = "Entrada"

    def percibir_y_actuar(self, habitacion, esta_sucia):
        self.habitacion_actual = habitacion
        print(f"[{self.nombre}] Inspeccionando habitación: {habitacion}")

        if self.nivel_bateria <= 10:
            print(f" --> Batería muy baja ({self.nivel_bateria}%). ¡Regresando a recargar!")
            return

        if esta_sucia:
            print(f" --> Se detectó suciedad. Limpiando {habitacion}...")
            self.nivel_bateria -= 10
            print(f" --> ¡Habitación limpia! Batería restante: {self.nivel_bateria}%")
        else:
            print(f" --> La habitación {habitacion} ya está limpia. Sin acción.")

# Código principal de prueba
robot = AgenteLimpiador("Robot-IA-1")

habitaciones = [
    ("Sala", True),
    ("Cocina", False),
    ("Dormitorio", True)
]

for habitacion, sucia in habitaciones:
    robot.percibir_y_actuar(habitacion, sucia)
```

```python
import random

class SensorTemperatura:
    def __init__(self, ubicacion):
        self.ubicacion = ubicacion
        self._temperatura = 20.0

    def leer_temperatura(self):
        # Simular variación de temperatura ambiental
        variacion = random.uniform(-1.5, 2.5)
        self._temperatura = round(self._temperatura + variacion, 1)
        return self._temperatura


class TermostatoInteligente:
    def __init__(self, umbral=25.0):
        self.umbral = umbral

    def evaluar_clima(self, temp_actual):
        print(f"Temperatura recibida: {temp_actual}°C")

        if temp_actual > self.umbral:
            print(" [ACTUADOR] ¡Atención! Temperatura alta. Encendiendo aire acondicionado.")
        else:
            print(" [ACTUADOR] Temperatura óptima. Apagando aire acondicionado.")


# Código principal de prueba
sensor = SensorTemperatura("Laboratorio de IA")
termostato = TermostatoInteligente(umbral=22.0)

for lectura in range(1, 4):
    print(f"\n--- Lectura #{lectura} ---")
    temp = sensor.leer_temperatura()
    termostato.evaluar_clima(temp)
```

```python
# Nueva prueba
sensor_prueba = SensorTemperatura("Salón de clases")
termostato_prueba = TermostatoInteligente(umbral=21.0)

for lectura in range(1, 5):
    print(f"\n--- Nueva lectura #{lectura} ---")
    temp = sensor_prueba.leer_temperatura()
    termostato_prueba.evaluar_clima(temp)
```

```python
import math

class Inmueble:
    def __init__(self, m2, cuartos, categoria="Desconocida"):
        self.m2 = m2
        self.cuartos = cuartos
        self.categoria = categoria

    def calcular_distancia(self, otro_inmueble):
        # Fórmula de distancia euclidiana
        d_m2 = (self.m2 - otro_inmueble.m2) ** 2
        d_cuartos = (self.cuartos - otro_inmueble.cuartos) ** 2
        return math.sqrt(d_m2 + d_cuartos)


# Casas con categoría conocida
casa_economica = Inmueble(m2=50, cuartos=2, categoria="Económica")
casa_lujo = Inmueble(m2=200, cuartos=5, categoria="Lujo")

# Casa nueva sin clasificar
casa_nueva = Inmueble(m2=60, cuartos=2)

dist_eco = casa_nueva.calcular_distancia(casa_economica)
dist_lujo = casa_nueva.calcular_distancia(casa_lujo)

print(f"Distancia a Casa Económica: {dist_eco:.2f}")
print(f"Distancia a Casa de Lujo: {dist_lujo:.2f}")

if dist_eco < dist_lujo:
    casa_nueva.categoria = casa_economica.categoria
else:
    casa_nueva.categoria = casa_lujo.categoria

print(f"--> La casa nueva se clasifica como: {casa_nueva.categoria}")
```

```python
class EstrategiaPromedio:
    def predecir(self, datos):
        # Predice usando la media aritmética
        return sum(datos) / len(datos)


class EstrategiaMaximo:
    def predecir(self, datos):
        # Predice tomando el valor máximo histórico
        return max(datos)


class PredictorIA:
    def __init__(self, estrategia):
        self.estrategia = estrategia

    def cambiar_estrategia(self, nueva_estrategia):
        self.estrategia = nueva_estrategia

    def realizar_prediccion(self, datos):
        return self.estrategia.predecir(datos)


# Código principal de prueba
temperaturas_semana = [22.0, 24.5, 21.0, 25.0, 23.5]

# Usar estrategia de promedio
predictor = PredictorIA(EstrategiaPromedio())
pred_prom = predictor.realizar_prediccion(temperaturas_semana)

print(f"Predicción (Estrategia Promedio): {pred_prom:.2f}°C")

# Cambiar a estrategia de máximo
predictor.cambiar_estrategia(EstrategiaMaximo())
pred_max = predictor.realizar_prediccion(temperaturas_semana)

print(f"Predicción (Estrategia Máxima): {pred_max:.2f}°C")
```

```python
# Nueva prueba con diferentes temperaturas
temperaturas_nuevas = [18.0, 20.5, 23.0, 19.5, 24.0]

predictor = PredictorIA(EstrategiaPromedio())

pred_prom = predictor.realizar_prediccion(temperaturas_nuevas)
print(f"Predicción (Estrategia Promedio): {pred_prom:.2f}°C")

predictor.cambiar_estrategia(EstrategiaMaximo())

pred_max = predictor.realizar_prediccion(temperaturas_nuevas)
print(f"Predicción (Estrategia Máxima): {pred_max:.2f}°C")
```

```python
class ModeloFiltroSpam:
    def ejecutar(self, entrada):
        return f"Analizando correo: '{entrada}' --> [Resultado: NO ES SPAM]"


class ModeloDeteccionRostros:
    def ejecutar(self, entrada):
        return f"Analizando imagen: '{entrada}' --> [Resultado: 2 ROSTROS DETECTADOS]"


class FabricaModelosIA:
    @staticmethod
    def crear_modelo(tipo_modelo):
        if tipo_modelo == "spam":
            return ModeloFiltroSpam()
        elif tipo_modelo == "rostros":
            return ModeloDeteccionRostros()
        else:
            raise ValueError(f"El modelo '{tipo_modelo}' no existe en el catálogo.")


# Código principal de prueba
modelo_1 = FabricaModelosIA.crear_modelo("spam")

print(modelo_1.ejecutar("¡Felicidades! Ganaste un premio gratis"))


modelo_2 = FabricaModelosIA.crear_modelo("rostros")

print(modelo_2.ejecutar("foto_graduacion.jpg"))
```

```python
class Pelicula:
    def __init__(self, titulo, genero):
        self.titulo = titulo
        self.genero = genero


class Usuario:
    def __init__(self, nombre, genero_favorito):
        self.nombre = nombre
        self.genero_favorito = genero_favorito


class SistemaRecomendacion:
    def __init__(self, catalogo):
        self.catalogo = catalogo

    def recomendar_para(self, usuario):
        print(f"Recomendaciones personalizadas para {usuario.nombre} "
              f"(Gusto: {usuario.genero_favorito}):")

        recomendaciones = []

        for peli in self.catalogo:
            if peli.genero == usuario.genero_favorito:
                recomendaciones.append(peli.titulo)

        return recomendaciones


# Código principal de prueba
catalogo = [
    Pelicula("Matrix", "Ciencia Ficción"),
    Pelicula("Star Wars", "Ciencia Ficción"),
    Pelicula("Orgullo y Prejuicio", "Romance"),
    Pelicula("Interestelar", "Ciencia Ficción")
]

usuario_ana = Usuario("Ana", "Ciencia Ficción")

sistema = SistemaRecomendacion(catalogo)

sugerencias = sistema.recomendar_para(usuario_ana)

for i, titulo in enumerate(sugerencias, 1):
    print(f" {i}. {titulo}")
```

```python
# Nuevo usuario con otro género favorito
usuario_nuevo = Usuario("Carlos", "Romance")

sugerencias_nuevas = sistema.recomendar_para(usuario_nuevo)

for i, titulo in enumerate(sugerencias_nuevas, 1):
    print(f" {i}. {titulo}")
```

```python
class Perceptron:
    def __init__(self, pesos, sesgo):
        self.pesos = pesos
        self.sesgo = sesgo

    def _funcion_activacion(self, suma_ponderada):
        # Función de paso simple
        return 1 if suma_ponderada >= 0 else 0

    def calcular_salida(self, entradas):
        # Multiplicación entrada * peso + sesgo
        suma = sum(e * p for e, p in zip(entradas, self.pesos)) + self.sesgo
        print(f" Suma ponderada calculada: {suma:.2f}")
        return self._funcion_activacion(suma)


# Código principal de prueba
# Neurona configurada con 2 entradas
neurona_aprobacion = Perceptron(
    pesos=[0.6, 0.4],
    sesgo=-6.0
)

# Calificaciones del estudiante
entradas_estudiante = [7.0, 8.0]

print("Evaluando aprobación del estudiante:")

resultado = neurona_aprobacion.calcular_salida(entradas_estudiante)

if resultado == 1:
    print("--> Decisión Neuronal: APROBADO (1)")
else:
    print("--> Decisión Neuronal: REPROBADO (0)")
```

```python
# Nueva prueba con otras calificaciones
entradas_nuevas = [5.0, 6.0]

print("Evaluando nuevas calificaciones:")

resultado_nuevo = neurona_aprobacion.calcular_salida(entradas_nuevas)

if resultado_nuevo == 1:
    print("--> Decisión Neuronal: APROBADO (1)")
else:
    print("--> Decisión Neuronal: REPROBADO (0)")
```

```python
class TransformadorBase:
    def __init__(self, nombre):
        self.nombre = nombre

    def fit_transform(self, datos):
        raise NotImplementedError(
            "Las clases hijas deben implementar este método."
        )


class EscaladorMinMax(TransformadorBase):
    def __init__(self):
        super().__init__("Escalador Min-Max (0 a 1)")

    def fit_transform(self, datos):
        val_min = min(datos)
        val_max = max(datos)

        print(f"[{self.nombre}] Mínimo: {val_min}, Máximo: {val_max}")

        # Aplicar fórmula a cada dato
        datos_escalados = [
            (x - val_min) / (val_max - val_min)
            for x in datos
        ]

        return [round(v, 3) for v in datos_escalados]


# Código principal de prueba
salarios = [1000, 2500, 5000, 10000]

escalador = EscaladorMinMax()

print("Datos originales de salarios:", salarios)

salarios_normalizados = escalador.fit_transform(salarios)

print("Salarios normalizados (0-1):", salarios_normalizados)
```

```python
# Nueva prueba con diferentes valores
precios = [500, 1200, 2500, 4000]

print("Datos originales de precios:", precios)

precios_normalizados = escalador.fit_transform(precios)

print("Precios normalizados (0-1):", precios_normalizados)
```

```python
import time


class ModeloInferenciaLenta:
    def predecir_sentimiento(self, texto):
        # Simular retardo computacional de un modelo profundo
        time.sleep(1.0)

        return "POSITIVO" if "excelente" in texto.lower() else "NEUTRO"


class ModeloConCache:
    def __init__(self):
        self.modelo_base = ModeloInferenciaLenta()
        self._cache = {}

    def predecir(self, texto):
        if texto in self._cache:
            print(" [CACHÉ] ¡Resultado encontrado en memoria instantánea!")
            return self._cache[texto]

        print(" [MODELO] Calculando predicción compleja (Demorado)...")

        resultado = self.modelo_base.predecir_sentimiento(texto)

        self._cache[texto] = resultado

        return resultado


# Código principal de prueba
ia_optimizada = ModeloConCache()

frase = "Este servicio es excelente"


print("--- Consulta 1 (Primera vez) ---")

t0 = time.time()
res1 = ia_optimizada.predecir(frase)

print(f"Resultado: {res1} | Tiempo: {time.time() - t0:.3f} s")


print("\n--- Consulta 2 (Misma frase repetida) ---")

t0 = time.time()
res2 = ia_optimizada.predecir(frase)

print(f"Resultado: {res2} | Tiempo: {time.time() - t0:.3f} s")
```

```python
# Nueva prueba con otra frase
frase_nueva = "El producto es excelente"

print("--- Nueva consulta ---")

t0 = time.time()
resultado_nuevo = ia_optimizada.predecir(frase_nueva)

print(f"Resultado: {resultado_nuevo} | Tiempo: {time.time() - t0:.3f} s")


print("\n--- Repetición de la consulta ---")

t0 = time.time()
resultado_repetido = ia_optimizada.predecir(frase_nueva)

print(f"Resultado: {resultado_repetido} | Tiempo: {time.time() - t0:.3f} s")
```
